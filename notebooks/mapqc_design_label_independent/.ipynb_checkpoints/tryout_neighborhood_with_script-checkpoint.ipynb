## Analysis:

For exploration, subset to a single cell type (specify matching annotation level as well). Eventually, we should do this on the entire embedding.

```python
ct1 = "AT2"  # AT2: level 3; DC2: level 4; Dendritic cells: level 3; Fibroblast lineage: level 2
ct1_level = 3
```

#### Query healthy:

(We should probably adapt and do healthy and disease at the same time, just separate results at the end)

```python
query_dataset = "Sheppard_2020"
```

```python
adata_ext_h = (
    adata_ext_full[
        [
            (ref_ann == ct1 or ext_ann == ct1)
            and (c_or_e == "core" or (cond == "IPF" and ds == query_dataset))
            for ref_ann, ext_ann, c_or_e, cond, ds in zip(
                adata_ext_full.obs[f"ann_level_{ct1_level}"],
                adata_ext_full.obs[f"transf_ann_level_{ct1_level}_label"],
                adata_ext_full.obs.core_or_extension,
                adata_ext_full.obs.lung_condition,
                adata_ext_full.obs.dataset,
            )
        ],
        :,
    ]
    .to_memory()
    .copy()
)
```

```python
adata_ext_h.obs.dataset.cat.categories
```

```python
len(adata_ext_h)
```

```python
adata_ext_h[adata_ext_h.obs.dataset == query_dataset].obs.lung_condition
```

```python
for k in [50, 500, 700, 1000, 1500, 1700]:
    sc.pp.neighbors(adata_ext_h, n_neighbors=k)
    prop_cells = 500
    for i in range(0, 100):
        res, neighborhood_stats = get_pvalues(adata_ext_h, k, prop_cells)
        neighborhood_stats.to_csv(f"C:/Users/49176/PycharmProjects/Bachelor/HLCA/New metric results/Sheppard_2020_IPF_AT2_progression/pvalues_{k}_{prop_cells}_{i}.csv")
```
