### Compare different n_hood_sizes and n_center_cells within Sheppard (in Fibroblasts)

```python
def get_stats(path):
    list_stats = []

    for k in [50, 500, 1000, 2000, 3000, 3500]:
        for prop_cells in range(50, k + 200, int(k / 10)):
            list_number = []
            list_fraction = []
            for i in range(0, 10):
                neighborhood_stats = pd.read_csv(
                    f"{path}\\pvalues_{k}_{prop_cells}_{i}.csv"
                )
                significant = len(
                    neighborhood_stats[neighborhood_stats.pvalue_corr < 0.05]
                )
                n_all = len(neighborhood_stats.dropna(inplace=False))
                list_number.append(significant)
                list_fraction.append(significant / n_all)
            list_stats.append(
                (
                    k,
                    1 / prop_cells,
                    np.mean(list_number),
                    np.std(list_number),
                    np.mean(list_fraction),
                    np.std(list_fraction),
                )
            )
    return pd.DataFrame(
        data=list_stats,
        columns=[
            "k",
            "prop_center_cells",
            "n_significant",
            "std_n_significant",
            "fraction_significant",
            "std_fraction",
        ],
    )
```

```python
def get_stats_2(path):
    list_stats = []

    for k in [50, 500, 700, 1000, 1500, 1700, 2000]:
        prop_cells = 500
        list_number = []
        list_fraction = []
        try:
            for i in range(0, 100):
                neighborhood_stats = pd.read_csv(
                    f"{path}\\pvalues_{k}_{prop_cells}_{i}.csv"
                )
                significant = len(
                    neighborhood_stats[neighborhood_stats.pvalue_corr < 0.05]
                )
                n_all = len(neighborhood_stats.dropna(inplace=False))
                list_number.append(significant)
                list_fraction.append(significant / n_all)
            list_stats.append(
                (
                    k,
                    1 / prop_cells,
                    np.mean(list_number),
                    np.std(list_number),
                    np.mean(list_fraction),
                    np.std(list_fraction),
                )
            )
        except:
            continue
    return pd.DataFrame(
        data=list_stats,
        columns=[
            "k",
            "prop_center_cells",
            "n_significant",
            "std_n_significant",
            "fraction_significant",
            "std_fraction",
        ],
    )
```

```python
def get_significant(path):
    list_sign = []
    for k in [50, 500, 1000, 2000, 3000, 3500]:
        tmp_list = []
        for prop_cells in range(50, k + 200, int(k / 10)):
            for i in range(0, 10):
                neighborhood_stats = pd.read_csv(
                    f"{path}\\pvalues_{k}_{prop_cells}_{i}.csv"
                )
                significant = neighborhood_stats[
                    neighborhood_stats.pvalue_corr <= 1
                ].pvalue_corr
                tmp_list.extend(significant)
        list_sign.append((k, tmp_list))
    df = pd.DataFrame(data=list_sign, columns=["k", "p_value"])
    return df.explode("p_value")
```

```python
def get_significant_2(path):
    list_sign = []
    for k in [50, 500, 700, 1000, 1500, 1700, 2000]:
        tmp_list = []
        prop_cells = 500
        try:
            for i in range(0, 100):
                neighborhood_stats = pd.read_csv(
                    f"{path}\\pvalues_{k}_{prop_cells}_{i}.csv"
                )
                significant = neighborhood_stats[
                    neighborhood_stats.pvalue_corr <= 1
                ].pvalue_corr
                tmp_list.extend(significant)
            list_sign.append((k, tmp_list))
        except:
            continue

    df = pd.DataFrame(data=list_sign, columns=["k", "p_value"])
    return df.explode("p_value")
```
