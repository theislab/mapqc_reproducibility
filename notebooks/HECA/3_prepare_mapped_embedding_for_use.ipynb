```python
import scanpy as sc
```

```python
%load_ext lab_black # for pretty code formatting
```

```python
adata_emb = sc.read_h5ad(
    "../../data/HECA/queries_mapped/Asherman_syndrome_mapped_emb.h5ad"
)
```

```python
# remove disease from reference, as we need healthy control reference only
adata_emb = adata_emb[adata_emb.obs.Group != "Endo_Superficial", :].copy()
```

recalculate neighbor graph and umap:

```python
sc.pp.neighbors(adata_emb, n_neighbors=30)
```

```python
sc.tl.umap(adata_emb)
```

```python
sc.tl.leiden(adata_emb, resolution=1)
```

clean up barcodes (remove final suffices, for some reason -0 and -1 etc. were added to all barcodes)

```python
new_barcodes = ["-".join(bc.split("-")[:-1]) for bc in adata_emb.obs.index]
```

```python
adata_emb.obs.index = new_barcodes
```

### Store result:

```python
adata_emb.write("../../data/HECA/queries_mapped/Asherman_syndrome_mapped_emb_subsetted_and_processed.h5ad")
```
