```python
atlas_path = "../../data/HLCA_full_v1.1_emb.h5ad"
dir_out = "./umaps/"
```

### Load data and subset:

```python
adata = sc.read_h5ad(atlas_path)
```

```python
query_datasets_actually_ref = [
    "Banovich_Kropski_2020",  # success with healthy, lack of mixing with disease
    "Misharin_Budinger_2018",  # success but not mixing so well in UMAP, good argument for quantitative metric
]
```

```python
print('test')
```

```python
query_datasets_q = [
    "Budinger_2020",
    "Duong_lungMAP_unpubl",
    "Eils_2020",  # nucleus fail
    "MeyerNikolic_unpubl_UCL",  # Healthy retry?
    "Meyer_2021_3prime",  # Healthy retry?
    "Schultze_unpubl",  # Healthy successful?
    "Sheppard_2020",  # Healthy + disease retry?
    "Tata_unpubl",  # Healthy
    "Thienpont_2018_10Xv2",  # Healthy re-try?
    "Zhang_2021",  # BALF (healthy + disease)
]
```

```python
for q_ds in query_datasets_actually_ref + query_datasets_q:
    print(q_ds)
    adata_subset = adata[
        (adata.obs.core_or_extension == "core") | (adata.obs.dataset == q_ds)
    ].copy()
    print(adata_subset.shape)
    # re-do neighbor graph
    sc.pp.neighbors(adata_subset,n_neighbors=30)
    # recalculate UMAP:
    sc.tl.umap(adata_subset)
    # and store
    umap_df = pd.DataFrame(adata_subset.obsm["X_umap"], index=adata_subset.obs.index)
    umap_df.to_csv(f"{dir_out}{q_ds}_umap.csv")
    print("\n")
```
