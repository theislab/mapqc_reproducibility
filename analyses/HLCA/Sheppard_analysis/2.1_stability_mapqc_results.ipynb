```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
%load_ext lab_black
```

```python
l = 60
```

```python
path_results_dir = "/lustre/groups/ml01/code/lisa.sikkema/atlas_extension/results/HLCA/Sheppard_2020/cellxgene_counts_batch_sample_kl1.0/mapqc/"
```

```python
res_control = pd.read_csv(
    os.path.join(path_results_dir, f"control_mapqc_stats_6_runs_l{l}.csv"), index_col=0
)
res_control["case_or_control"] = "control"
res_case = pd.read_csv(
    os.path.join(path_results_dir, f"case_mapqc_stats_6_runs_l{l}.csv"), index_col=0
)
res_case["case_or_control"] = "case"
```

```python
res = res_control.merge(res_case, how="outer")
```

```python
res
```

```python
fig, ax = plt.subplots()
sns.boxplot(y=res.perc_non_mixing_cells, x=res.case_or_control, ax=ax)
sns.stripplot(y=res.perc_non_mixing_cells, x=res.case_or_control, ax=ax)
plt.ylim(0, 100)
plt.show()
```
